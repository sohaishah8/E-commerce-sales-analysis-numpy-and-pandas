```python
import numpy as np
import pandas as pd
df=pd.read_csv('ecommerce_10000_customers_analysis_practice.csv')
print('===rows,columns===:\n',df.shape)
print('===columns===:\n',df.columns,'\n')
print('====information===:\n\n',df.info)
print('mean==medain===mode:\n',df.describe())
print('====missing order======\n\n\n------------\n',df.isnull().sum(),'\n')
print('===% of missing order======:\n\n',df.isnull().sum()/len(df)*100)
print('====dublicate value if present====:\n\n',df.duplicated().sum())
print('===remove duplicates value===\n\n',df.drop_duplicates())
df['Discount']=df['Discount'].fillna(0) #CLAENING DATA
df['Country']=df['Country'].fillna('unknown')
df['Payment_Method']=df['Payment_Method'].fillna('unknown')
print('is value clean or not\n------------------------\n',df.isnull().sum())
print('checking claning\n-----------------------\n',df.duplicated().sum())
df['Order_Date']=pd.to_datetime(df['Order_Date'])
print() #have to ask.
df['year']=df['Order_Date'].dt.year
df['month']=df['Order_Date'].dt.month
#sales_analysis.
total_sales=df['Sales'].sum()
average_sales=df['Sales'].mean()
maximum_sales=df['Sales'].max()
print('total_sales:',total_sales)
print('===average_sales===:\n\n',average_sales)
print('===highest order sales===:\n\n',maximum_sales)
#quantity analysis
total_quantity=df['Quantity'].sum()
average_quantity=df['Quantity'].mean()
highest_quantity=df['Quantity'].max()
print('===total quantity:====\n',total_quantity)
print('===highest quantity===\n',highest_quantity)
print('===average quantity===\n',average_quantity)
#category_analysis
category_sales=df.groupby('Category')['Sales'].sum()
print('===category sales:\n=====\n',category_sales)
#category analysis
category_analysis=df.groupby('Category').agg(
    total_sales=('Sales','sum'),
    total_quantity=('Quantity','sum'),
    number_of_orders=('Order_ID','count')
)
print('===the category of sales is:\n========\n',category_analysis)
#product analysis
product_sales=df.groupby('Product')['Sales'].sum()
print('====product sales====:\n',product_sales)
product_analysis=df.groupby('Product').agg(
    total_sales=('Sales','sum'),
    total_Quantity=('Quantity','sum'),
    numbers_of_orders=('Order_ID','count')
)
print('====product analysis=====:\n',product_analysis)
#customer analysis
customer_analysis=df.groupby('Customer_ID').agg(
    total_sales=('Sales','sum'),
    total_quantity=('Quantity','sum'),
    number_of_orders=('Order_ID','count')
)
print('=====the total customer analysis=====\n',customer_analysis)
#srif top 02 customer niklne ha..
print(customer_analysis.sort_values('total_sales',ascending=False).head(2))
#country analysis
country_analysis=df.groupby('Country').agg(
    total_sales=('Sales','sum'),
    total_quantity=('Quantity','sum'),
    number_of_orders=('Order_ID','count')
)
print('=====country analysis=====:\n',country_analysis)
#top 02 sales....
print(country_analysis.sort_values('total_sales',ascending=False).head(2))
#numpy analysis
sales=df['Sales'].dropna().to_numpy()
print('average sales:',np.mean(sales))
print('medain:',np.median(sales))
print('standerd deviation:',np.std(sales))
print('maximam sales:',np.max(sales))
print('minimum sales:',np.min(sales))
#final summery 
print('=====E-commerce sales Report======')
print('total sales:',df['Sales'].sum())
print('average sales:',df['Sales'].mean())
print('total quantity sales:',df['Quantity'].sum())
print('total order:',df['Order_ID'].unique())
Top_category=df.groupby('Category')['Sales'].sum().sort_values(ascending=False).index[0]
Top_product=df.groupby('Product')['Sales'].sum().sort_values(ascending=False).index[0]
Top_country=df.groupby('Country')['Sales'].sum().sort_values(ascending=False).index[0]
print('Top category:\n',Top_category)
print('Top product:\n',Top_product)
print('top country:\n',Top_country)
```
